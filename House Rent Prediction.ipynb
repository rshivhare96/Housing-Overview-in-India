## 🏠 Housing Overview in India

India’s housing landscape is incredibly diverse, ranging from the grand palaces of former maharajas to compact huts in remote villages, and modern apartment complexes in metropolitan areas. With the rise in household incomes, the housing sector in India has seen significant expansion. According to the Human Rights Measurement Initiative, India currently achieves **60.9% of the potential** for fulfilling the right to housing based on its income level.

## 📊 Dataset Overview

This dataset contains detailed information on **4,700+ residential properties** available for rent across various cities in India. Each listing includes several parameters to help understand the housing options better:

### Dataset Features

- **📌 BHK** – Number of Bedrooms, Hall, and Kitchen  
- **📌 Rent** – Rental cost of the property  
- **📌 Size** – Area of the property in square feet  
- **📌 Floor** – Floor number where the property is located and the total number of floors in the building  
- **📌 Area Type** – Type of area measurement used (e.g., Super built-up, Built-up, Carpet area)  
- **📌 Area Locality** – Specific locality or neighborhood of the property  
- **📌 City** – City where the property is located  
- **📌 Furnishing Status** – Whether the property is furnished, semi-furnished, or unfurnished  
- **📌 Tenant Preferred** – Preferred type of tenant (e.g., family, bachelor, company)  
- **📌 Bathroom** – Number of bathrooms in the property  
- **📌 Point of Contact** – Contact information for inquiries related to the property

Bar Plot for Number of House in Each City which is Available for Rent


```python
sns.set_context("poster", font_scale = .8)
plt.figure(figsize = (20, 6))
ax = rent_df["City"].value_counts().plot(kind = 'bar',
                                         color = "crimson",
                                         rot = 0)

for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + 0.25, p.get_height() - 100),
                ha = 'center',
                va = 'bottom',
                color = 'white')
```

Bar Plot on Different Types of Furnishing Status

```python
sns.set_context("poster", font_scale = .8)
plt.figure(figsize = (20, 6))
ax = rent_df["Furnishing Status"].value_counts().plot(kind = 'bar',
                                                      color = "Orange",
                                                      rot = 0)

for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + 0.25, p.get_height() - 250),
                ha = 'center',
                va = 'bottom',
                color = 'Black')
```

Bar Plot on Different Types of Tenant Preferred

```python
sns.set_context("poster", font_scale = .8)
plt.figure(figsize = (20, 6))
ax = rent_df["Tenant Preferred"].value_counts().plot(kind = 'bar', color = "Green", rot = 0)

for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + 0.25, p.get_height() - 300),
                ha = 'center',
                va = 'bottom',
                color = 'White')
```

Pie Plot on Cities to check the distribution

```python
plt.figure(figsize = (20, 8))
counts = rent_df["City"].value_counts()
explode = (0, 0, 0, 0, 0, 0.1)
colors = ['#FF1E00', '#A66CFF', '#EAE509', '#D61C4E', '#3CCF4E', '#3AB4F2']

counts.plot(kind = 'pie',
            colors = colors,
            explode = explode,
            autopct = '%1.1f%%')
plt.axis('equal')
plt.legend(labels = counts.index, loc = "best")
plt.show()
```
